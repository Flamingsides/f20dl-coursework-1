## Model Training

This next section trains our model based on the recommended split in the dataset

```python
SPLIT = "is_train" # Use recommended 50-50 split
META_COLS = ["class_id", "image_id", "class_name", "is_train", "is_train_80"]

# Split training and testing data
train, test = df[df[SPLIT] == 1], df[df[SPLIT] == 0]

# Get all feature columns
feature_cols = [col for col in df.columns if col not in META_COLS]
X_train, y_train = train[feature_cols], train.class_id
X_test, y_test = test[feature_cols], test.class_id

print("X_train.shape: ", X_train.shape)
print("X_test.shape: ", X_test.shape)
print("y_train.shape: ", y_train.shape)
print("y_test.shape: ", y_test.shape)
```

### Begin training

Code inspired from Lecture 12 Google Collab. We use the default lbfgs solver.

```python
# Initialise and train model
log_reg = LogisticRegression(penalty="l2", tol=1e-08, max_iter=1000)
```

```python
log_reg.fit(X_train, y_train)
```

```python
y_preds_train = log_reg.predict(X_train)
y_preds_test = log_reg.predict(X_test)
from sklearn.metrics import accuracy_score

print(f"Model accuracy (training set): {accuracy_score(y_train, y_preds_train) * 100:.4f}%")
print(f"Model accuracy (test set): {accuracy_score(y_test, y_preds_test) * 100:.4f}%")
```

The model accuracy on the test set is quite low, so we could try to improve it through feature selection techniques. The training set is still high so it could be overfitting.

Let's first try a different solver. According to tests by [Hale, J. (2019)](https://dev.to/discdiver/don-t-sweat-the-solver-stuff-20np), liblinear performs slightly better for multi-class datasets.

```python
# Initialise and train model
log_reg_1 = LogisticRegression(solver="liblinear", penalty="l2", tol=1e-08, max_iter=1000)

log_reg_1.fit(X_train, y_train)

y_preds_train = log_reg_1.predict(X_train)
y_preds_test = log_reg_1.predict(X_test)
from sklearn.metrics import accuracy_score

print(f"Model accuracy (training set): {accuracy_score(y_train, y_preds_train) * 100:.4f}%")
print(f"Model accuracy (test set): {accuracy_score(y_test, y_preds_test) * 100:.4f}%")
```

Funnily enough, our results reveal that the original lbfgs solver performed better. Let's try to understand our features better and perform some feature analysis.

```python
# Calculate correlation using Pearson's r.

from sklearn.feature_selection import r_regression

X, y = df[feature_cols], df.class_id
pearsons_r_all = r_regression(X, y)

# Map feature columns' names to respective Pearson's r
zipped = list(zip(feature_cols, pearsons_r_all))
pearsons_r_per_col = pd.DataFrame(zipped, columns=["feature", "pearsons_r"])

pearsons_r_per_col = pearsons_r_per_col.sort_values(axis=0, by=["pearsons_r"], key=lambda x: -abs(x))

pearsons_r_per_col.head(10)
```

The above test reveals that even the most correlating feature only has a pearson's r of 0.24 which means no attribute strongly correlates with the class_id individually. We can instead check if any combination of features improves the correlation.

```python
X, y = df[feature_cols].copy(), df.class_id.copy()

X["has_any_yellow"] = X[[col for col in feature_cols if "yellow" in col]].aggregate("max", axis=1)

pearsons_r_has_any_yellow = r_regression(X[["has_any_yellow"]], y)
pearsons_r_has_any_yellow
```

An even lower correlation! However, at this point, it has been noted that Pearson's r doesn't make sense here since the class ids are arbritrary. Other methods like mutual_info_classif and chi2 are more appropriate (recommendation by Claude).

Code inspired from:
1. [sklearn.feature_selection.mutual_info_classif](https://scikit-learn.org/stable/modules/generated/sklearn.feature_selection.mutual_info_classif.html)
1. [sklearn.feature_selection.chi2](https://scikit-learn.org/stable/modules/generated/sklearn.feature_selection.chi2.html)

```python
from sklearn.feature_selection import mutual_info_classif, chi2

mutual_info = mutual_info_classif(X, y)

# Map feature columns' names to respective mutual info rating
zipped = list(zip(feature_cols, mutual_info))
mutual_info_per_col = pd.DataFrame(zipped, columns=["feature", "mutual_info"])
mutual_info_per_col = mutual_info_per_col.sort_values(axis=0, by=["mutual_info"], key=lambda x: -abs(x))

mutual_info_per_col.head(10)
```

```python
chi2_stats, p_vals = chi2(X, y)

zipped = list(zip(feature_cols, chi2_stats, p_vals))
chi2_per_col = pd.DataFrame(zipped, columns=["feature", "chi2_stats", "p_vals"])
chi2_per_col = chi2_per_col.sort_values(axis=0, by=["chi2_stats", "p_vals"], key=lambda x: -abs(x))

chi2_per_col.head(200)
```

It seems that most of the features are strongly influential to the target. Thus, the issue is perhaps overfitting. Let's then try to train once again while we change the regularisation term to make it general better. Slide 27 of lecture 10 slides says that L1 is useful for irrelevant, redundant features. It may help in our case since we have low mutual_info scores.

```python
# Initialise and train model
log_reg_2 = LogisticRegression(solver="liblinear", penalty="l2", tol=1e-08, max_iter=1000)

log_reg_2.fit(X_train, y_train)

y_preds_train = log_reg_2.predict(X_train)
y_preds_test = log_reg_2.predict(X_test)
from sklearn.metrics import accuracy_score

print(f"Model accuracy (training set): {accuracy_score(y_train, y_preds_train) * 100:.4f}%")
print(f"Model accuracy (test set): {accuracy_score(y_test, y_preds_test) * 100:.4f}%")
```

The accuracy is now even lower :(

```python
# Switch back to lbfgs solver
log_reg_3 = LogisticRegression(penalty="l2", tol=1e-08, max_iter=1000)

high_chi2_columns = chi2_per_col[chi2_per_col["chi2_stats"] > 1000]["feature"].to_list()

log_reg_3.fit(X_train[high_chi2_columns], y_train)

y_preds_train = log_reg_3.predict(X_train[high_chi2_columns])
y_preds_test = log_reg_3.predict(X_test[high_chi2_columns])
from sklearn.metrics import accuracy_score

print(f"Model accuracy (training set): {accuracy_score(y_train, y_preds_train) * 100:.4f}%")
print(f"Model accuracy (test set): {accuracy_score(y_test, y_preds_test) * 100:.4f}%")
```

It appears that the best accuracy we got was the first one with all features used, the lbfgs solver, and l2 penalty. The reason we can only get this far for our accuracy must be because of the complex fine-grained dataset that we are dealing with. This dataset is described as a fine-grained challenge in the associate paper. Perhaps we need some methods that are better at handling non-linear patterns.
